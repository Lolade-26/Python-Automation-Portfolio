```python
#Filtering Data

import pandas as pd

data = {
    "Name": ["John", "Mary", "Peter"],
    "Score": [80, 90, 75]
}

df = pd.DataFrame(data)
print(df)
```

```python
Filtered = df[df["Score"] >  80]
print(Filtered)
```

```python
#SCORE ANALYZER

df = pd.DataFrame(data)

print("Average Score:", df["Score"].mean())
print("Highest Score:", df["Score"].max())
print("Students above 85:")
print(df[df["Score"] > 85])
```
